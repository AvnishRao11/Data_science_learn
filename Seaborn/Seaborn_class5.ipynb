#  Seaborn – Class 5 & 6: Matrix/Grid Plots + Styling


## Why Matrix / Grid Plots?

Matrix or grid-style plots are perfect for:
- Showing correlations between variables  
- Visualizing pivot tables  
- Discovering patterns in large datasets


### Heatmaps (heatmap())

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
tips=sns.load_dataset('tips')
sns.heatmap(tips.corr(numeric_only=True),annot=True)
```

### Cluster Map (clustermap())

```python
iris=sns.load_dataset('iris')
sns.clustermap(iris.corr(numeric_only=True),annot=True)
```

### Styling & Themes

```python
sns.set_style('darkgrid') #options : white, dark , darkgrid, whitegrid , ticks
sns.set_palette('rainbow') #options : deep, muted, pastel, bright, dark, colorblind
# sns.set_context('notebook') #options : paper, notebook, talk, poster
sns.scatterplot(x="total_bill",y='tip',data=tips,hue='sex',s=80)
```
